Árbol de decisión con el dataset Iris. Este cuaderno muestra  un ejemplo práctico y sencillo de cómo utilizar un árbol de decisión en python utilizando la biblioteca Scikitlearn. Nuestro objetivo es aprende a clasificar datos usando árboles de decisión

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt
```

```python
iris = load_iris()
X = iris.data #caracteristicas
y = iris. target #etiquetas
```

#Dividimos los datos en entrenamiento y pruebas

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

Creamos el model de arbo de decisión

```python
modelo = DecisionTreeClassifier()
modelo.fit(X_train, y_train)
```

Hacemos predicciones

```python
y_pred = modelo.predict(X_test)
```

evaluamos el modelo

```python
precision = accuracy_score(y_test, y_pred)
print(f"Precisión del modelo: {precision:.2f}")
```

```python
#Visualización del árbol
plt.figure(figsize=(12, 8))
plot_tree(modelo, feature_names=iris.feature_names, class_names=iris.target_names, filled=True, rounded=True)
plt.show()
```
